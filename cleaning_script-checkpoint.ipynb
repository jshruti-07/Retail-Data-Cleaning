```

import pandas as pd

# Load raw dataset
df = pd.read_csv("retail_raw.csv")

# Remove duplicates
df.drop_duplicates(inplace=True)

# Fix missing values
df["Price"] = df["Price"].fillna(df["Price"].median())

# Convert Date column
df["Date"] = pd.to_datetime(df["Date"])

# Remove outliers in Quantity
Q1 = df["Quantity"].quantile(0.25)
Q3 = df["Quantity"].quantile(0.75)
IQR = Q3 - Q1
df_clean = df[(df["Quantity"] >= Q1 - 1.5*IQR) & (df["Quantity"] <= Q3 + 1.5*IQR)]

# Save cleaned file
df_clean.to_csv("retail_cleaned.csv", index=False)

df_clean.head()
```
